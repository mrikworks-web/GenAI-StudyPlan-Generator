```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

model_name = "Qwen/Qwen2.5-0.5B-Instruct"

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype=torch.float32
)

print("Model loaded successfully!")
```

```python
messages = [
    {
        "role": "system",
        "content": (
            "You are an academic study planner. "
            "Follow the student's constraints exactly. "
            "Do not invent syllabus topics."
        )
    },
    {
        "role": "user",
        "content": """
Create a 3-day study plan with these details:

Subjects: Mathematics and Python
Available study time: 3 hours per day
Difficult subject: Mathematics
Student level: Intermediate
Goal: Exam preparation and concept revision

Requirements:
- Schedule exactly 3 hours of study per day.
- Use three 50-minute study sessions.
- Include a 10-minute break between sessions.
- Give Mathematics more study time than Python.
- Include revision and practice questions.
- Use only general study activities, not invented syllabus topics.
- Clearly label each day.
- Keep the response concise and complete.
"""
    }
]

text = tokenizer.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=True
)

inputs = tokenizer(text, return_tensors="pt")

outputs = model.generate(
    **inputs,
    max_new_tokens=400,
    do_sample=False,
    repetition_penalty=1.1
)

response = tokenizer.decode(
    outputs[0][inputs["input_ids"].shape[1]:],
    skip_special_tokens=True
)

print(response)
```

```python
# Student Information

print("=== PERSONALIZED STUDY PLAN GENERATOR ===\n")

name = input("Enter your name: ")

subjects = input(
    "Enter your subjects (separate with commas): "
).split(",")

subjects = [subject.strip() for subject in subjects]

study_hours = float(
    input("How many hours can you study per day? ")
)

exam_days = int(
    input("How many days are left until your exams? ")
)

difficulty = input(
    "Which subject do you find the most difficult? "
)

learning_style = input(
    "How do you prefer to study? "
    "(e.g., videos, reading, practice questions): "
)

goal = input(
    "What is your main study goal? "
)

print("\n=== STUDENT PROFILE ===")
print("Name:", name)
print("Subjects:", subjects)
print("Daily study hours:", study_hours)
print("Days until exams:", exam_days)
print("Most difficult subject:", difficulty)
print("Learning preference:", learning_style)
print("Study goal:", goal)
```

```python
messages = [
    {
        "role": "system",
        "content": """
You are a professional academic study planner.

Rules:
- Use only facts provided by the student.
- Never invent the student's progress, strengths, or weaknesses.
- Do not make assumptions about gender.
- Explain recommendations using the student's actual inputs.
- Follow the requested format.
- If information is missing, give general advice instead of guessing.
"""
    },
    {
        "role": "user",
        "content": f"""
Create a personalized study strategy using this student profile.

Subjects: {", ".join(subjects)}
Daily available study time: {study_hours} hours
Days until exams: {exam_days}
Most difficult subject: {difficulty}
Learning preference: {learning_style}
Goal: {goal}

Follow this exact structure:

1. Subject Priorities
Briefly explain how to allocate attention across the subjects.
Give extra attention to the difficult subject.

2. Learning Techniques
Suggest 3 techniques suited to the stated learning preference.

3. Revision Strategy
Explain how to use the available {exam_days} days.
Include an initial learning phase, regular revision, and final review.
Do not shorten or change the number of available days.

4. Practical Recommendations
Give 3 actionable recommendations.

Do not invent syllabus topics or claim the student has already
completed any preparation.

Keep the answer concise and complete.
"""
    }
]

text = tokenizer.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=True
)

inputs = tokenizer(text, return_tensors="pt")

outputs = model.generate(
    **inputs,
    max_new_tokens=600,
    do_sample=False,
    repetition_penalty=1.15
)

recommendations = tokenizer.decode(
    outputs[0][inputs["input_ids"].shape[1]:],
    skip_special_tokens=True
)

print("=== PERSONALIZED STUDY RECOMMENDATIONS ===\n")
print(recommendations)
```

```python
# Personalized Timetable Generator

import math

def generate_timetable(subjects, study_hours, exam_days, difficulty):
    total_minutes = int(study_hours * 60)
    session_length = 50
    break_length = 10

    difficult_subject = difficulty.strip().lower()

    # Find the difficult subject in the subject list
    hard_subject = next(
        (s for s in subjects if s.lower() == difficult_subject),
        subjects[0]
    )

    other_subjects = [
        s for s in subjects if s.lower() != difficult_subject
    ]

    if not other_subjects:
        other_subjects = [hard_subject]

    # Create study sessions
    session_count = math.ceil(total_minutes / session_length)

    print("=== YOUR PERSONALIZED STUDY TIMETABLE ===\n")

    for day in range(1, exam_days + 1):
        print(f"\nDAY {day}")
        print("-" * 35)

        remaining = total_minutes
        elapsed = 0

        for session in range(session_count):
            duration = min(session_length, remaining)
            if duration <= 0:
                break

            # Prioritize the difficult subject in alternating sessions
            if session % 2 == 0:
                subject = hard_subject
            else:
                index = ((day - 1) + session // 2) % len(other_subjects)
                subject = other_subjects[index]

            start = elapsed
            end = elapsed + duration

            def format_time(minutes):
                hour = 9 + minutes // 60
                minute = minutes % 60
                return f"{hour:02d}:{minute:02d}"

            print(
                f"{format_time(start)} - {format_time(end)} | "
                f"{subject} ({duration} min)"
            )

            elapsed = end
            remaining -= duration

            if remaining > 0:
                elapsed += break_length
                print(f"Break ({break_length} min)")

        print(f"Total study time: {total_minutes // 60}h "
              f"{total_minutes % 60}m")

generate_timetable(
    subjects=subjects,
    study_hours=study_hours,
    exam_days=exam_days,
    difficulty=difficulty
)
```
